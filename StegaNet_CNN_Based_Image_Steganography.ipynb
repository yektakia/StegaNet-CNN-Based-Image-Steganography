```python
# -------------------------
# CNN-Based Image Steganography (Colab-ready)
# -------------------------

# 1️⃣ Install dependencies
!pip install scikit-image opencv-python tqdm

# 2️⃣ Imports
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
import os
from PIL import Image
from google.colab import files

# 3️⃣ Device
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# -------------------------
# 4️⃣ Hyperparameters
# -------------------------
data_dir = './data'
output_dir = './outputs'
epochs = 5
batch_size = 32
prep_ch = 32
mid_ch = 64
size = 32
beta = 1.0
log_interval = 100

os.makedirs(output_dir, exist_ok=True)

# -------------------------
# 5️⃣ Dataset (CIFAR-10)
# -------------------------
transform = transforms.Compose([
    transforms.Resize((size, size)),
    transforms.ToTensor(),
])

train_dataset = torchvision.datasets.CIFAR10(root=data_dir, train=True, download=True, transform=transform)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)

# -------------------------
# 6️⃣ CNN Models
# -------------------------
class Encoder(nn.Module):
    def __init__(self, in_ch=3, prep_ch=prep_ch, mid_ch=mid_ch):
        super().__init__()
        self.prep = nn.Conv2d(in_ch, prep_ch, kernel_size=3, padding=1)
        self.conv1 = nn.Conv2d(prep_ch + in_ch, mid_ch, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(mid_ch, mid_ch, kernel_size=3, padding=1)
        self.out = nn.Conv2d(mid_ch, in_ch, kernel_size=3, padding=1)
        self.relu = nn.ReLU()

    def forward(self, cover, secret):
        cover_p = self.relu(self.prep(cover))
        x = torch.cat([cover_p, secret], dim=1)
        x = self.relu(self.conv1(x))
        x = self.relu(self.conv2(x))
        stego = torch.sigmoid(self.out(x))
        return stego

class Decoder(nn.Module):
    def __init__(self, in_ch=3, mid_ch=mid_ch):
        super().__init__()
        self.conv1 = nn.Conv2d(in_ch, mid_ch, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(mid_ch, mid_ch, kernel_size=3, padding=1)
        self.out = nn.Conv2d(mid_ch, in_ch, kernel_size=3, padding=1)
        self.relu = nn.ReLU()

    def forward(self, stego):
        x = self.relu(self.conv1(stego))
        x = self.relu(self.conv2(x))
        secret_hat = torch.sigmoid(self.out(x))
        return secret_hat

encoder = Encoder().to(device)
decoder = Decoder().to(device)

# -------------------------
# 7️⃣ Loss and optimizer
# -------------------------
criterion = nn.MSELoss()
optimizer = optim.Adam(list(encoder.parameters()) + list(decoder.parameters()), lr=1e-3)

# -------------------------
# 8️⃣ Training loop
# -------------------------
for epoch in range(epochs):
    encoder.train()
    decoder.train()
    running_loss = 0.0
    for batch_idx, (cover, _) in enumerate(tqdm(train_loader)):
        cover = cover.to(device)
        # Randomly pick secret images
        secret_idx = torch.randint(0, len(train_dataset), (cover.size(0),))
        secret = torch.stack([train_dataset[i][0] for i in secret_idx]).to(device)

        optimizer.zero_grad()
        stego = encoder(cover, secret)
        secret_hat = decoder(stego)
        loss = criterion(stego, cover) + beta * criterion(secret_hat, secret)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

        if (batch_idx + 1) % log_interval == 0:
            print(f"Epoch [{epoch+1}/{epochs}], Batch [{batch_idx+1}], Loss: {running_loss/log_interval:.4f}")
            running_loss = 0.0

    # Visualization
    encoder.eval()
    decoder.eval()
    with torch.no_grad():
        sample_cover = train_dataset[0][0].unsqueeze(0).to(device)
        sample_secret = train_dataset[1][0].unsqueeze(0).to(device)
        stego_img = encoder(sample_cover, sample_secret)
        revealed_secret = decoder(stego_img)

        def to_np(t):
            return t.squeeze(0).permute(1,2,0).cpu().numpy()

        fig, axes = plt.subplots(1,4, figsize=(12,3))
        axes[0].imshow(to_np(sample_cover)); axes[0].set_title("Cover"); axes[0].axis('off')
        axes[1].imshow(to_np(sample_secret)); axes[1].set_title("Secret"); axes[1].axis('off')
        axes[2].imshow(to_np(stego_img)); axes[2].set_title("Stego"); axes[2].axis('off')
        axes[3].imshow(to_np(revealed_secret)); axes[3].set_title("Revealed"); axes[3].axis('off')
        plt.show()

print("✅ Training complete!")

# -------------------------
# 9️⃣ Interactive hiding demo
# -------------------------
def hide_and_download():
    print("Upload your cover image:")
    uploaded_cover = files.upload()
    cover_path = list(uploaded_cover.keys())[0]

    print("Upload your secret image:")
    uploaded_secret = files.upload()
    secret_path = list(uploaded_secret.keys())[0]

    transform = transforms.Compose([
        transforms.Resize((size, size)),
        transforms.ToTensor()
    ])

    cover_img = transform(Image.open(cover_path).convert('RGB')).unsqueeze(0).to(device)
    secret_img = transform(Image.open(secret_path).convert('RGB')).unsqueeze(0).to(device)

    encoder.eval()
    with torch.no_grad():
        stego_img = encoder(cover_img, secret_img)

    stego_pil = transforms.ToPILImage()(stego_img.squeeze(0).cpu())
    stego_pil.save("stego_result.png")
    print("Stego image saved as stego_result.png")
    files.download("stego_result.png")

# Run interactive demo
hide_and_download()
```
